```python
# function for the addition of two numbers
def add(a, b):
  return a + b
```

```python
# funtion for the subtraction of a number from another
def subtract(a, b):
  return a - b
```

```python
# function for the multiplication of two numbers
def multiply(a, b):
  return a * b
```

```python
# function for the division of a number by another
def divide(a, b):
  return a / b
```

```python
def calculate(choice, x, y):
  if choice == 'add':
    return f'The result of {x} + {y} is {add(x, y)}'
  elif choice == 'subtract':
    return f'The result of {x} - {y} is {subtract(x, y)}'
  elif choice == 'multiply':
    return f'The result of {x} * {y} is {multiply(x, y)}'
  elif choice == 'divide':
    return f'The result of {x} / {y} is {divide(x, y)}'
  else:
    return 'Invalid choice'
```

```python
print(calculate('add', 1, 2))
print(calculate('subtract', 1, 2))
print(calculate('multiply', 1, 2))
print(calculate('divide', 1, 2))
```
